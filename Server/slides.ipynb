```python
import uuid
def flatten_tree(node, parent_id=None, flat_list=None):
    if flat_list is None:
        flat_list = []

    map_id = str(uuid.uuid4())
    node_dict = {
        'map_id': map_id,
        'map_parent': parent_id,
        'title': node['keyword'],
        'content': node['description'],
        'class': 'ce907bb8-f51e-4933-b9a2-d042c5b05e67',
        'lectures': node.get('lectures', [])
    }
    flat_list.append(node_dict)
    for child in node.get('children', []):
        flatten_tree(child, parent_id=map_id, flat_list=flat_list)
    return flat_list

flat_list = flatten_tree(output)
```
