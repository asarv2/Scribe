```python
# pulling from supabase to get chapters and homeworks

from dotenv import load_dotenv
load_dotenv()

import os
from supabase.client import Client, create_client

supabase_url = os.getenv("SUPABASE_URL")
supabase_key = os.getenv("SUPABASE_PRIVATE_KEY")
supabase: Client = create_client(supabase_url, supabase_key)
```

```python
# defining the textbook ids
v_textbook_id = "c12f6017-008d-4ab0-8398-3d03e64d6169"
mg_textbook_id = "bb79425d-af45-4cdc-8bef-c62461a7ee48"
c_textbook_id = "de868f53-f8d1-4c3f-bacf-bbfbb3303d0c"
```

```python
# getting textbooks
v_textbook = supabase.table("textbooks").select("*").eq("id", v_textbook_id).execute()
mg_textbook = supabase.table("textbooks").select("*").eq("id", mg_textbook_id).execute()
c_textbook = supabase.table("textbooks").select("*").eq("id", c_textbook_id).execute()

v_textbook = v_textbook.data[0]
mg_textbook = mg_textbook.data[0]
c_textbook = c_textbook.data[0]

print(v_textbook)
print(mg_textbook)
print(c_textbook)
```

```python
# defining page offsets for each of the textbooks, which can differ at different intervals
# each tuple defines (start_page, offset)
v_page_offset = [(0, 22), (27, 21), (123, 20), (149, 19), (171, 18), (187, 17), (215, 16), (275, 15), (295, 14), (309, 13), (333, 12), (351, 11), (389, 10), (433, 9)]
mg_page_offset = [(0, 8), (29, 7), (81, 6), (193, 5)]
c_page_offset = [(0, 16)]
```

```python
from PyPDF2 import PdfReader, PdfWriter
# read in the pdfs for each of the textbooks
V = PdfReader(open("V.pdf", "rb"))
MG = PdfReader(open("MG.pdf", "rb"))
C = PdfReader(open("C.pdf", "rb"))
```

### Chapters

```python
# getting the pages of each of the chapters
v_chapters = supabase.table("chapters").select("*").eq("textbook", v_textbook_id).order("page_number").execute()
mg_chapters = supabase.table("chapters").select("*").eq("textbook", mg_textbook_id).order("page_number").execute()
c_chapters = supabase.table("chapters").select("*").eq("textbook", c_textbook_id).order("page_number").execute()

print(v_chapters)
print(mg_chapters)
print(c_chapters)
```

```python
# extracting information
v_chapters = v_chapters.data
mg_chapters = mg_chapters.data
c_chapters = c_chapters.data

print(v_chapters)
print(mg_chapters)
print(c_chapters)
```

```python
def get_chapter_pages(chapters, page_offsets, end_page):
    # generate tuples of (start_page, end_page) for each chapter
    chapter_pages = []
    for i in range(len(chapters)):
        page_num = chapters[i]['page_number']
        
        # determine offset based on page range
        offset = 0
        for start, off in page_offsets:
            if page_num >= start:
                offset = off
                
        start_page = page_num + offset
        
        if i != len(chapters) - 1:
            next_page = chapters[i+1]['page_number']
            # get offset for next page
            next_offset = 0
            for start, off in page_offsets:
                if next_page >= start:
                    next_offset = off
            end_page_chapter = next_page + next_offset - 1
        else:
            # for last chapter, use end page variable
            end_page_chapter = end_page + offset
            
        chapter_pages.append((start_page, end_page_chapter))
    return chapter_pages

# Get chapter pages for each textbook
v_chapter_pages = get_chapter_pages(v_chapters, v_page_offset, v_textbook['pages'])
mg_chapter_pages = get_chapter_pages(mg_chapters, mg_page_offset, mg_textbook['pages'])
c_chapter_pages = get_chapter_pages(c_chapters, c_page_offset, c_textbook['pages'])

```

```python
def split_textbook_by_chapters(textbook_name, pdf_reader, chapters, chapter_pages):
    # create directory for the textbook
    os.makedirs(textbook_name, exist_ok=True)
    
    # split pdf by chapter pages and save to directory
    for i in range(len(chapters)):
        start_page, end_page = chapter_pages[i]
        output_path = f"{textbook_name}/Chapter {i+1} - {chapters[i]['title']}.pdf"
        
        with open(output_path, "wb") as f:
            writer = PdfWriter()
            for page in range(start_page - 1, end_page):
                writer.add_page(pdf_reader.pages[page])
            writer.write(f)

# Split each textbook
split_textbook_by_chapters("V", V, v_chapters, v_chapter_pages)
split_textbook_by_chapters("MG", MG, mg_chapters, mg_chapter_pages) 
split_textbook_by_chapters("C", C, c_chapters, c_chapter_pages)
```

### Homework

```python
# doing homework
homework = supabase.table("homework").select("*").execute()
homework = homework.data
print(homework)
```

```python
# get homework_problems for each homework
homework_problems = supabase.table("homework_problems").select("*").in_("homework", [hw['id'] for hw in homework]).execute()
homework_problems = homework_problems.data

# binning homework problems by assignment
assignments = {}
for hp in homework_problems:
    assignments[hp['homework']] = assignments.get(hp['homework'], []) + [hp]    
print(assignments)
```

```python
def get_mapped_pages(assignment_id, assignments):
    """
    Get the mapped page numbers for all problems in an assignment and save them as PDFs
    """
    assignment_problems = assignments[assignment_id]
    
    # Get the homework title for this assignment_id from the homework list
    homework_title = next((hw['title'] for hw in homework if hw['id'] == assignment_id), 'Untitled')
    
    # Create directory for this homework assignment
    assignment_dir = f"homework/{homework_title}"
    os.makedirs(assignment_dir, exist_ok=True)
    
    # for each assignment problem, get the page number, and find the mapped page number in the textbook using the page_offsets
    for ap in assignment_problems:
        page_number = ap['page_number']
        textbook = ap['textbook']
        
        if textbook == v_textbook_id:
            page_offset = v_page_offset
            pdf_reader = V
        elif textbook == mg_textbook_id:
            page_offset = mg_page_offset
            pdf_reader = MG
        else:
            page_offset = c_page_offset
            pdf_reader = C
        
        if page_number == None:
            # create text file with the problem number and additional info
            problem_number = ap['problem_number'] if ap['problem_number'] != None else ''
            output_path = f"{assignment_dir}/{problem_number}.txt"
            with open(output_path, "w") as f:
                f.write(ap['additional_info'] if ap['additional_info'] != None else '')
        else:
            offset = 0
            for start, off in page_offset:
                if page_number >= start:
                    offset = off
            
            mapped_page_number = page_number + offset - 1
                
            # Create PDF for this problem in the assignment directory
            problem_number = ap['problem_number'] if ap['problem_number'] != None else ''
            output_path = f"{assignment_dir}/{problem_number}.pdf"
            
            with open(output_path, "wb") as f:
                writer = PdfWriter()
                writer.add_page(pdf_reader.pages[mapped_page_number])
                # Add metadata with additional info
                writer.add_metadata({'/Info': ap['additional_info'] if ap['additional_info'] != None else ''})
                writer.write(f)
```

```python
# Example usage for all assignments
for assignment_id in assignments:
    get_mapped_pages(assignment_id, assignments)
```
