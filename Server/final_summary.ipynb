```python
from dotenv import load_dotenv
load_dotenv()
import os
from supabase.client import Client, create_client

supabase_url = os.getenv("SUPABASE_URL")
supabase_key = os.getenv("SUPABASE_PRIVATE_KEY")
supabase: Client = create_client(supabase_url, supabase_key)
```

```python
response = supabase.table("summaries").select("*").execute()
print(response)
```

```python
# getting all the valid slides
slide_response = supabase.table("slides").select("*").eq("class", "ce907bb8-f51e-4933-b9a2-d042c5b05e67").execute()
print(slide_response)
```

```python
# only getting the summaries that have slides
valid_summaries = [summary for summary in response.data if summary["slide"] in [slide["id"] for slide in slide_response.data]]
print(valid_summaries)
print(len(valid_summaries))
```

```python
prompt = """You are an expert professor tasked with creating a comprehensive lecture summary. For the lecture provided between <START> and <END> tags, generate a detailed and thorough summary that includes:

1. Lecture Overview
- Main topics and concepts covered
- Learning objectives and key takeaways
- Prerequisites and connections to other lectures

2. Detailed Content Summary
- In-depth explanation of each major concept
- Step-by-step breakdowns of important processes
- Clear examples illustrating each key point

3. Technical Components
- Definitions of all important terms
- Complete mathematical formulas and notations
- Theorems, proofs, and key equations
- Detailed explanation of each symbol and variable

4. Practical Applications
- Multiple real-world examples for each concept
- Industry use cases and applications
- Common implementation scenarios

5. Study Guide
- Common pitfalls and misconceptions to avoid
- Practice problems with detailed solutions
- Tips for mastering difficult concepts
- Review questions for self-assessment

Format the response using:
- Clear hierarchical structure with headings
- Bullet points for easy scanning
- Tables for comparing concepts
- Numbered steps for processes
- LaTeX for mathematical notation

Ensure each lecture summary is comprehensive, technically accurate, and written in an engaging, easy-to-understand style suitable for students. Maintain academic rigor while making complex concepts accessible.

Parse the lecture content using the following format:
<START Lecture Name>
[Lecture content]
<END>

Provide a complete and thorough summary for the lecture content."""
```

```python
# ask gemini to generate the summary
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model = 'gemini-1.5-flash-8b',
    temperature = 0,
    max_tokens = None,
    timeout = None,
    max_retries = 2
)

def get_slide_name(slide_id):
    # find the slide with the given id
    for slide in slide_response.data:
        if slide["id"] == slide_id:
            return slide["name"]
    return None

combined_prompt = prompt + "\n" + "\n".join([ "<START " + get_slide_name(summary["slide"]) + ">" + (summary["content"] + "\n") + "<END>" for summary in valid_summaries])

response = llm.invoke(combined_prompt)
print(response)
```
